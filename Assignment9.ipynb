## 1. Line Plot

This program creates a line plot using Seaborn.

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y = [10, 20, 15, 25, 30]

sns.lineplot(x=x, y=y, marker='o')

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Seaborn Line Plot")
plt.show()
```

## 2. Bar Plot

This program creates a bar plot using Seaborn to compare different categories.

```python
import seaborn as sns
import matplotlib.pyplot as plt

categories = ["A", "B", "C", "D", "E"]
values = [25, 40, 30, 50, 35]

sns.barplot(x=categories, y=values)

plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Seaborn Bar Plot")
plt.show()
```

## 3. Histogram

This program creates a histogram to show the distribution of numerical data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = [10, 20, 15, 25, 30, 20, 15, 35, 40, 25,
        30, 20, 15, 45, 35, 25, 30, 40, 20, 15]

sns.histplot(data, bins=5)

plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Seaborn Histogram")
plt.show()
```

## 4. Box Plot

This program creates a box plot to represent the distribution of numerical data.

```python
import seaborn as sns
import matplotlib.pyplot as plt

data = [10, 20, 15, 25, 30, 20, 15, 35, 40, 25,
        30, 20, 15, 45, 35, 25, 30, 40, 20, 15]

sns.boxplot(y=data)

plt.ylabel("Values")
plt.title("Seaborn Box Plot")
plt.show()
```

## 5. Scatter Plot

This program creates a scatter plot to show the relationship between two variables.

```python
import seaborn as sns
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5, 6]
y = [10, 15, 12, 20, 18, 25]

sns.scatterplot(x=x, y=y)

plt.xlabel("X Values")
plt.ylabel("Y Values")
plt.title("Seaborn Scatter Plot")
plt.show()
```
