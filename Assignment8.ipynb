## 1. Multi-Plot Figure Using Subplots

This program creates multiple plots in a single figure using `plt.subplot()`.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30]
y2 = [5, 15, 10, 20, 25]

plt.subplot(2, 1, 1)
plt.plot(x, y1)
plt.title("First Plot")

plt.subplot(2, 1, 2)
plt.plot(x, y2)
plt.title("Second Plot")

plt.tight_layout()
plt.show()
```

## 2. Styling, Legend and Grid

This program demonstrates styling of a plot along with a legend and grid.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]
y1 = [10, 20, 15, 25, 30]
y2 = [5, 15, 10, 20, 25]

plt.plot(x, y1, marker='o', linestyle='--', label="Data 1")
plt.plot(x, y2, marker='s', linestyle='-', label="Data 2")

plt.title("Styled Line Plot")
plt.legend()
plt.grid(True)

plt.show()
```

## 3. Multi-Plot with Legend and Grid

The following program creates multiple plots and applies titles, legends, grids, and styling.

```python
import matplotlib.pyplot as plt

x = [1, 2, 3, 4, 5]

# First Plot
plt.subplot(2, 2, 1)
plt.plot(x, [10, 20, 15, 25, 30], marker='o', label="Sales")
plt.title("Sales")
plt.legend()
plt.grid(True)

# Second Plot
plt.subplot(2, 2, 2)
plt.plot(x, [5, 15, 10, 20, 25], marker='s', label="Profit")
plt.title("Profit")
plt.legend()
plt.grid(True)

# Third Plot
plt.subplot(2, 2, 3)
plt.plot(x, [20, 25, 30, 35, 40], marker='^', label="Production")
plt.title("Production")
plt.legend()
plt.grid(True)

# Fourth Plot
plt.subplot(2, 2, 4)
plt.plot(x, [8, 12, 18, 22, 28], marker='*', label="Growth")
plt.title("Growth")
plt.legend()
plt.grid(True)

plt.tight_layout()
plt.show()
```
