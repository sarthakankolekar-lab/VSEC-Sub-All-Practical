## 1. Import Required Libraries

Import Pandas, Matplotlib, and Seaborn for data analysis and visualization.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

## 2. Load the Titanic Dataset

The Titanic dataset contains information about passengers such as age, sex, passenger class, fare, and survival status.

```python
df = sns.load_dataset("titanic")

df.head()
```

## 3. Understand the Dataset

Display the shape, column names, data types, and first few records of the dataset.

```python
print("Shape of dataset:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nDataset Information:")
df.info()
```

## 4. Summary Statistics

The `describe()` function provides statistical information about the numerical columns.

```python
df.describe()
```

## 5. Missing Values Detection

Check the number of missing values present in each column.

```python
missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

## 6. Missing Values Visualization

A bar plot is used to visualize the number of missing values in each column.

```python
missing_values = df.isnull().sum()

missing_values = missing_values[missing_values > 0]

plt.figure(figsize=(8, 5))
sns.barplot(x=missing_values.index, y=missing_values.values)

plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.title("Missing Values in Titanic Dataset")
plt.xticks(rotation=45)
plt.show()
```

## 7. Correlation Matrix

Calculate the correlation between numerical variables in the dataset.

```python
correlation = df.corr(numeric_only=True)

print(correlation)
```

## 8. Correlation Heatmap

The correlation matrix is represented graphically using a Seaborn heatmap.

```python
plt.figure(figsize=(10, 7))

sns.heatmap(correlation, annot=True, cmap="coolwarm", linewidths=0.5)

plt.title("Correlation Matrix - Titanic Dataset")
plt.show()
```

## 9. Visual Analysis

Visualize the relationship between passenger class, fare, and survival using Seaborn.

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="class", y="survived")

plt.xlabel("Passenger Class")
plt.ylabel("Average Survival")
plt.title("Survival Rate by Passenger Class")
plt.show()
```
